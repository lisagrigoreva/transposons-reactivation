```python
## Load data 
import pandas as pd 
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
te_expression=pd.read_csv("/groups/nordborg/projects/ddm1_resilencing/04_output/limma_output/accession_specific_effect/expression_salmon_limma_normilized_tes_signif_padj001_log2fc1_accession_union_wo_controls.csv", sep=",",decimal=",")  
te_expression
```

#### Remove TEs which overlap genes

```python
## Remove overapped TEs with genes 
## Filter out TEs based on the overlap with genes 
overlap_gene_id=pd.read_csv("/groups/nordborg/projects/ddm1_resilencing/04_output/te_filtering_overlap/filtered_atleast1bp_overalp_with_gene_id_list.csv", sep=",",decimal=",", header=None) 
# Filter overlapping genes 
te_expression= te_expression[~te_expression['Unnamed: 0'].isin(overlap_gene_id[0])]
#del te_expression['Unnamed: 0']
te_expression = te_expression.set_index('Unnamed: 0')
te_expression.index.name = 'TE_ID'
```

```python
te_expression
```

```python
te_expression = te_expression.apply(pd.to_numeric, errors='coerce')
ddm1_matrix = te_expression.loc[:, te_expression.columns.str.contains("ddm1")]
col_matrix = te_expression.loc[:, te_expression.columns.str.contains("Col")]
# Transpose ddm1 and col0 matricies 
ddm1_matrix = ddm1_matrix.transpose()
col_matrix = col_matrix.transpose()
ddm1_matrix_corr=ddm1_matrix.corr(method="pearson")
col_matrix_corr=col_matrix.corr(method="pearson")
```

```python
## Make a clustering based on the TE
from scipy.cluster.hierarchy import linkage, dendrogram
import seaborn as sns
import matplotlib.pyplot as plt

# Compute the linkage matrix
linkage_matrix = linkage(ddm1_matrix, method='ward')

# Create a clustered heatmap
plt.figure(figsize=(12, 8))
sns.clustermap(ddm1_matrix,
               cmap='YlOrRd',  # or any other colormap
               yticklabels=False,
               xticklabels=False,
               figsize=(12, 8),metric="euclidean")
plt.title('Clustered TE Expression')
plt.show()
```

```python
from scipy.spatial import distance
from scipy.cluster import hierarchy

correlations = ddm1_matrix.corr()
correlations_array = np.asarray(ddm1_matrix.corr())

row_linkage = hierarchy.linkage(
    distance.pdist(correlations_array), method='average')

col_linkage = hierarchy.linkage(
    distance.pdist(correlations_array.T), method='average')

sns.clustermap(correlations, row_linkage=row_linkage, col_linkage=col_linkage, method="average", figsize=(13, 13))
```

```python
import pandas as pd
import numpy as np

def extract_pairwise_correlations(corr_matrix):
    # Get column names (TE IDs)
    te_ids = corr_matrix.columns

    # Create empty list to store correlations
    correlations = []

    # Iterate through upper triangle
    for i in range(len(te_ids)):
        for j in range(i + 1, len(te_ids)):
            correlations.append({
                'TE1': te_ids[i],
                'TE2': te_ids[j],
                'Correlation': corr_matrix.iloc[i, j]
            })

    # Convert to DataFrame
    result = pd.DataFrame(correlations)

    # Sort by absolute correlation
    result['Abs_Correlation'] = abs(result['Correlation'])
    result = result.sort_values('Abs_Correlation', ascending=False)

    return result[['TE1', 'TE2', 'Correlation']]

# Example usage:
pairwise_correlations = extract_pairwise_correlations(ddm1_matrix_corr)

# Save to CSV if needed
# pairwise_correlations.to_csv('pairwise_correlations.csv', index=False)
```

```python
## Get highest correlations 
filtered_correlation=pairwise_correlations[(pairwise_correlations['Correlation']>0.8)]
## Write dataframe 
filtered_correlation
```
